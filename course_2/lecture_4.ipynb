## Time Bars

**Time bars** are created by summarizing price and trading activity over fixed time intervals of equal length, such as every minute, every hour, or every day. Instead of recording every individual trade, all activity that occurs within each interval is compressed into a single summary record, called a *bar* (also commonly known as a *candle* or *candlestick*).

Each time bar typically contains the following information, all referring to the same fixed time interval:

* **Timestamp**: the time at which the interval ends (or, by convention, begins).
* **Open price**: the first price at which a trade occurred during the interval.
* **Close price**: the last price at which a trade occurred during the interval.
* **High price**: the highest price reached during the interval.
* **Low price**: the lowest price reached during the interval.
* **Volume traded**: the total number of units (shares, contracts, etc.) traded during the interval.
* **Volume-Weighted Average Price (VWAP)**: the average price during the interval, where each trade price is weighted by the size of the trade, so that prices at which more volume was traded have a greater influence on the average.

Although time bars are widely used by both practitioners and researchers, they are problematic for two main reasons.

First, **markets do not evolve at a constant pace over time**. Trading activity is much more intense during certain periods (such as market open or major news releases) and much quieter during others. Fixed-length time bars ignore these differences in market speed. As a result, bars created during calm periods may contain very little information, while bars created during active periods may compress a large amount of information into the same time span. This means that time bars are **not directly comparable**, even though they all cover the same amount of clock time.

Second, this uneven flow of information leads to **unstable and distorted statistical behavior** in time-based price series. For example, price changes may appear artificially clustered or erratic simply because some bars represent slow markets and others represent fast markets. In plain terms, the data can look noisier or more irregular than it truly is, making modeling and forecasting more difficult. Many statistical techniques have been developed to correct for these issues, but the root cause is the choice of time-based sampling itself.

To illustrate this point, imagine a market in which the trading speed (or market speed) is constant, but the time intervals over which the data (prices and volumes) are aggregated vary randomly. For example, we might first aggregate over 47 minutes, then over 361 minutes, then over 130 minutes, and so on. Suppose further that, when we present the resulting aggregated bars, we provide no information about the length of the aggregation period. It is easy to see that such bars would be of limited practical value. This situation closely resembles the case under consideration: although the time intervals are fixed, the market’s effective speed varies over time, and we neither observe this speed nor account for it in the aggregation process.

A more effective approach is to construct bars based on **trading activity rather than clock time**, for example, forming a new bar after a certain amount of volume or number of trades has occurred. By doing so, each bar represents a similar level of market activity, which avoids these problems from the outset.

## Tick Bars

**Tick bars** are built by summarizing market activity after a **fixed number of transactions**, rather than after a fixed amount of time. A **tick** simply means a single recorded transaction or trade. For example, if we choose a tick size of 1,000, a new bar is created every time 1,000 trades have occurred, regardless of how long that takes in clock time.

As with time bars, each tick bar summarizes all activity within that group of trades and typically includes values such as the timestamp, open price, close price, high price, low price, traded volume, and the volume-weighted average price (VWAP).

The main advantage of tick bars is that they **align data sampling with trading activity**, which is a practical proxy for how quickly new information is entering the market. When markets are active, ticks occur rapidly and bars are formed frequently. When markets are quiet, ticks arrive slowly and bars take longer to form. This naturally adapts the data to the market’s changing speed.

Sampling prices based on a fixed number of transactions tends to produce **more stable and predictable statistical behavior** than sampling based on fixed time intervals. In simple terms, price changes measured over equal amounts of trading activity tend to behave more consistently than price changes measured over equal amounts of clock time. This consistency is important because many statistical tools assume that data points are:

* **Independent** (one observation does not depend on the previous one),
* **Identically distributed** (they follow the same statistical pattern over time),
* And roughly **normally distributed** (values cluster around an average in a symmetric way).

When these conditions are approximately satisfied, statistical models can make more reliable conclusions. This is also called *drawing inference*: using observed data to make meaningful statements, estimates, or predictions about market behavior. For inference to be reliable, the underlying behavior of the data should remain **stable over time**, meaning its statistical properties should not change simply because the market becomes faster or slower. Tick bars help achieve this stability by ensuring that each bar represents a similar level of market activity.

When constructing tick bars, it is important to handle **outliers** carefully. Many exchanges use **auctions** at the market open and close. In this context, an auction is a trading mechanism where buy and sell orders are collected over a short period without immediately matching them. During this accumulation phase, the **order book** (the list of outstanding buy (bids) and sell (offers) orders) fills up, but no trades occur.

When the auction ends, all compatible orders are matched at a single **clearing price**, which is the price that allows the maximum number of shares or contracts to be traded. This results in one very large transaction being reported. Even though it appears as a single tick, this trade may represent the equivalent of thousands of normal trades. The traded quantity in such a case is said to be **outsized**, meaning unusually large compared to typical transactions.

If not treated properly, these auction trades can distort tick bars by concentrating an abnormally large amount of activity into what appears to be just one transaction.

## Volume Bars

A key limitation of **tick bars** is that **counting transactions is not always a reliable measure of market activity**, because transactions can vary greatly in size. In other words, two tick bars containing the same number of trades may represent very different amounts of actual trading.

This problem arises because of **order fragmentation**, which refers to how buy and sell interest is split into individual orders. For example, imagine that there is a single sell order offering 10 units at a given price. If a buyer purchases all 10 units at once, this will be recorded as **one transaction**, or one tick. However, if those same 10 units are offered as ten separate sell orders of 1 unit each, the exact same purchase will be recorded as **ten separate transactions**, even though the economic outcome is identical.

In addition, the systems that operate exchanges (often called **matching engines**) may split what is logically a single trade into several smaller reported transactions. Sometimes, for technical or operational reasons, one order execution is reported as multiple **partial fills**, each representing a fraction of the total traded quantity (a **fill** refers to the execution of an order, meaning the moment when a buy order and a sell order are matched). These partial fills are “artificial” in the sense that they do not reflect separate trading decisions, but rather how the system processes and reports trades.

Because of these effects, simply counting ticks can introduce unnecessary randomness into tick bars. **Volume bars** avoid this issue by summarizing market activity every time a **fixed amount of volume** has traded, regardless of how many individual transactions were involved. For example, instead of forming a bar after 1,000 trades, we might form a bar every time 1,000 shares or contracts have changed hands.

By construction, volume bars ensure that each bar represents the same amount of traded quantity, making them a more consistent measure of market activity than time bars or tick bars. This often leads to **more stable and well-behaved price changes**, which is useful for statistical analysis and modeling.

## Dollar Bars

**Dollar bars** are constructed by creating a new bar every time a **fixed amount of money has been traded**. In simpler terms, instead of counting time, ticks, or volume, we form a bar whenever the *total price of the traded instruments* reaches a chosen dollar amount, such as $1 million.

The motivation for dollar bars becomes clear when prices change significantly over time. Suppose the price of a stock **doubles** over a certain period. At the beginning, investing 1,000 dollars requires buying many shares; at the end, investing the same 1,000 dollers requires buying only half as many shares. This shows that the **number of shares traded depends on the price**, while the actual economic activity is better captured by the amount of money exchanged.

This aligns with how traders think in practice. Traders usually do not decide how many shares to buy or sell in isolation. Instead, they think in terms of **how much money they are willing to invest or risk**. Because of this, sampling market data based on dollar value traded provides a more consistent representation of trading activity than sampling based on time, number of trades, or raw volume (especially when prices fluctuate a lot).

This difference has clear practical consequences. When bars are constructed using time, tick count, or volume, the **number of bars per day can vary dramatically over time**, because market activity and prices change. When bars are constructed using dollar value, this variation tends to be much smaller. In other words, dollar bars adapt naturally to changes in both trading intensity and price levels, producing a more stable stream of observations.

Another important advantage of dollar bars relates to changes in the **number of outstanding shares** of a company. The number of outstanding shares represents how many shares of the company exist and can be traded in the market. This number can change over time due to corporate actions such as issuing new shares. When the number of outstanding shares changes, measures based on ticks or volume can become misleading, because the same level of economic activity may now produce very different trade counts or volumes. Dollar bars are less affected by these changes because they focus on **economic value traded**, not on how many units were exchanged. For this reason, dollar bars tend to remain meaningful even as the structure of the company evolves.

In some cases, it is beneficial to **adjust the dollar bar size dynamically** instead of keeping it fixed forever. As a company grows or shrinks, the total value of its shares available for trading (often measured by its **free-floating market capitalization**) also changes. If the bar size remains fixed while the company becomes much larger or smaller, the bars may become too frequent or too sparse. Adjusting the bar size in line with the company’s market size helps ensure that each bar continues to represent a comparable level of market activity over time.
