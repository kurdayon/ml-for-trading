## 1. Introduction

In this chapter, I provide a high-level overview of trading and machine learning, as well as an explanation of how the two are related. By “high-level”, I mean that the discussion intentionally avoids deep dives into nuances and technical details.

## 2. Trading

There is a wide variety of so-called financial instruments. Examples include foreign currencies, cryptocurrencies such as Bitcoin, stocks, bonds, treasuries, and more advanced instruments like options and futures. Despite their differences, all of these instruments share a common property: they can be bought and sold relatively easily at a price that changes over time. By trading these instruments intelligently, it is possible to earn money. For instance, if the price of an instrument increases after you purchase it, you can sell it at a higher price and realize a profit. When this process is carried out regularly with the goal of generating profit, it is referred to as trading. 

Trading is not an easy way to earn money because the future movement of prices is unknown. For example, if we expect the price of an instrument to increase, we might buy it with the intention of selling it later at a higher price. However, it is entirely possible that after we buy the instrument, the price moves in the direction opposite to the expected one (falls). In that case, we incur a loss. This means that, to be a successful trader, one must be able to predict the future at least to some extent. Roughly speaking, it is not necessary to be right all the time, but it is essential to be right more often than wrong.

When trading is not performed manually but instead executed automatically by a computer following a predefined (programmed) set of rules, it is referred to as **algorithmic trading**. The term emphasizes that trading decisions and executions are handled by an algorithm rather than a human trader.

Another important aspect of trading is how decisions are made. A trader may base decisions on intuition or subjective judgment, reasoning through each situation individually. In contrast, trading decisions can also be made by consistently applying the same well-defined rules that specify when to trade (buy or sell) and in what quantity. This approach is known as **systematic trading**, as it relies on a fixed decision-making system rather than discretion.

In this sense, any form of algorithmic trading is also a type of systematic trading. While all algorithmic trading is systematic, not all systematic trading is algorithmic. A trader could, in principle, follow a strict rule-based system manually without automation.

## 3. Machine Learning

Machine learning (ML) is a field of science concerned with discovering patterns (dependencies) in noisy data. Consider a situation in which we observe a set of inputs that influence an outcome in a way that is unknown to us. A concrete example helps illustrate this idea.

Suppose the prices of apartments are the outcomes, determined by various inputs such as the size of the apartment, its age, distance from the city center, number of rooms, proximity to a subway station, and other factors. We know that these inputs affect the price, but we do not know exactly how. The relationships may be nonlinear, and the inputs may interact with one another. For example, the effect of the year of construction might differ for apartments located closer to the city center than for those farther away.

To uncover such complex relationships between inputs and outcomes, we need to collect many examples in the form of input–output pairs and provide them to a machine learning algorithm. The algorithm (machine) learns from the data and produces a function that maps inputs to outputs. Once learned, this function can be evaluated on new inputs. As a result of this evaluation you get a prediction of the apartments price.

This picture should always be complemented by the fact that the dependency between inputs and outcomes is probabilistic (stochastic, noisy). In other words, the inputs do not determine the outcome deterministically; instead, they influence the probabilities of different possible outcomes. For example, for a given set of inputs, some price may be very unlikely, while for others inputs this price may be highly probable. Consequently, even with exactly the same inputs, different outcomes may be observed. As a result, even a perfect model cannot be expected to produce predictions that always exactly match the true values we aim to predict.

## 4. Financial Machine Learning

So, how are systematic trading and machine learning related? As mentioned earlier, earning money from trading requires the ability to predict future price movements in order to make trading decisions, specifically, when to buy, when to sell, and in what quantity.

One way to predict future price movements is through experience. By carefully observing price dynamics over time, a trader may detect recurring patterns. For example, experience might reveal that prices tend to increase more in certain situations than in others. A situation can be described by the values of various factors, often referred to as parameters, inputs, features, indicators, or signals. To make this idea more concrete, such factors might include the day of the week, the percentage of positive news related to a country’s financial situation, or characteristics of recent price movements of a financial instrument.

At this point, you can already see how the pieces of the puzzle fit together. To predict future price movements, we need to identify patterns in data—and machine learning is a field of science that provides methods for discovering such patternsin a rigorous, scientific and close to optimal way.

More specifically, we can collect many examples in the form of input–output pairs. As inputs, we use factors that we believe may predict future price movements. As outputs, we use the actual price changes observed after those inputs were recorded. With a sufficient number of such examples, we can apply machine learning techniques to train a model that learns how the outputs (future price movements) depend on the inputs.

There are several reasons why machine learning can identify patterns more effectively than humans. Humans typically reason in terms of relatively simple, discrete rules and linear relationships involving a small number of inputs. In contrast, machines can detect highly complex, nonlinear patterns that depend on a large number of inputs.

Moreover, machines can process billions of data points in a relatively short period of time. A huge amount of data is a curse for a person but a blessing for a machine. Machines are also better at detecting very weak patterns that are obscured by large amounts of noise. Finally, machine learning methodology helps distinguish genuine patterns from spurious ones (so-called phantoms) that arise purely from noise. In other words, humans may fail to detect real patterns that machines can uncover, while at the same time perceiving patterns where none actually exist. Machine learning systems, by contrast, are better equipped to identify true structure in data and to classify random fluctuations as noise.

Can you engage in systematic trading without machine learning? Let’s think this through.

Systematic trading, by definition, means following a set of rules that determine how you trade (when to buy, when to sell, how much to trade, and which financial instruments to use). Any such rule necessarily relies on patterns. For example, a trading strategy (your rule) might detect a pattern (a specific relationship among inputs) that signals a favorable moment to buy.

But where do these patterns come from? The only reliable source of patterns is the real world or, in other words, our experience captured in data. These patterns emerge from observing how markets have behaved in the past. This is precisely what machine learning is designed to do: analyze large amounts of historical data to uncover meaningful patterns, and machine learning does this far more efficiently and systematically than manual or ad hoc approaches, as we have already discussed.

From this perspective, machine learning is not optional for a systematic trader, it is essential! If your goal is to identify patterns, you should use the most powerful tools available, developed over decades by some of the brightest minds specifically for that purpose: searching for patterns in data.

## 5. Challenges of Financial Machine Learning

Machine learning provides a general, field-agnostic methodology: the same core models and techniques can be applied across diverse domains such as retail, medicine, astrophysics, biology, or marketing. For example, neural networks are used in many disciplines, and the fundamental steps involved in building, training, and evaluating them are largely the same regardless of the application.

Finance, however, represents a notable exception. The challenges encountered when applying machine learning to financial data and financial problems (such as designing effective training strategies) are sufficiently unique that they have given rise to the term "Financial Machine Learning". This term reflects the reality that standard machine learning approaches often require significant modification and extension to be effective in financial settings. In this section, we provide a brief overview of the key reasons why this is the case. In this book, we directly address these challenges and offer practical, actionable solutions.

### 5.1. High Noise-to-Signal Ratio and Overfitting

In machine learning datasets, the values of the outputs (also known as *labels* or *targets*) are assumed to be a function of the corresponding inputs, plus an additional random component known as *noise*.

We can illustrate this using the considered earlier example of apartment prices. We assume that the price of an apartment can be computed by some function of observable inputs associated with the apartment, such as its size, age, and distance from the city center. However, the actual price of each apartment will deviate randomly from this computed value. This deviation can be attributed to factors that are not included among the input variables. For example, the price may be influenced by the quality of the view from windows, the urgency with which the owner needs to sell, or even the owner’s mood at the time the price is set.

The level of noise depends on the problem under consideration. For example, in our example, it may be that the function of the available inputs explains 90% of the price, while the remaining 10% of the observed price is a random or unpredictable contribution arising from factors unknown to us.

In domains where noise dominates (such as finance) the situation is very different. There, the overwhelming majority of price movements may be driven by noise, say 99%, with only a very small fraction, remaining 1%, attributable to the dependency captured by our function. We refer to this potentially predictable component as the *signal*. We say that in such cases the noise-to-signal ratio is very high.

A very high noise-to-signal ratio has a clear underlying cause: market efficiency. What does this mean? Financial markets consist of a vast number of experienced, knowledgeable, and intelligent participants. These traders buy an instrument as soon as it appears underpriced, that is, when its price is lower than it should be, and sell when it appears overpriced. As a result, prices are continuously pushed toward their “fair” values. Consequently, an instrument is almost never meaningfully under- or overpriced. If it becomes even slightly underpriced, market participants quickly start buying it, driving the price up. This mechanism implies that we should rarely observe situations in which a price can be expected to clearly rise in the near future, because such an expectation would itself indicate underpricing. In short, if markets were perfectly efficient, there would be no exploitable patterns, no signal at all, only random noise. If markets are slightly inefficient, then small signals may exist, but they are overwhelmed by noise, leading to a very high noise-to-signal ratio.

High noise presents a major challenge for machine learning. If there were no noise in the data, many machine-learning problems would reduce to simple optimization tasks: we would merely need to find a function that fits the data perfectly. In other words, we would search for a function that returns the observed outputs (for example, apartment prices) for the corresponding inputs (the properties of the apartments).

When noise is present, however, the situation changes fundamentally. The function we seek is obscured (or hidden) by noise, making it much harder to identify the underlying relationship between inputs and outputs.

An intuitive way to understand this is through a visual analogy. An image is simply a collection of pixels with different colors. If you randomly alter the color of each pixel by a small amount, the image remains recognizable, but it appears noisier and of lower quality. As the level of random alteration increases, it becomes progressively harder to discern what the image depicts. At very high noise levels, the image disappears entirely and all that remains is noise, much like the image on an old television screen when there is no signal.

A similar phenomenon might occurs when listening to the radio. When the signal is weak, the audio becomes difficult to understand. When noise dominates, it becomes nearly impossible to extract any meaningful information from the transmission.

In the example above, we showed that when inputs are extremely noisy, a machine-learning model (like a human observer) may be unable to recognize the true patterns hidden beneath the noise. There is, however, a closely related problem, the other side of the coin. Because an ML model is explicitly designed to search for signal (that is, patterns or dependencies), it may try too hard to find structure in the data and, as a result, it can mistake noise for signal and end up identifying “fake” patterns that do not actually exist. The same phenomenon occurs in human perception. When looking at random spots, we may see faces; when listening to noisy sounds, we may hear voices. People can also perceive patterns in sequences of real-world events where no genuine pattern is present. This confusion of noise with signal, and the consequent detection of nonexistent, false patterns, is known as **overfitting**.

A significant portion of machine learning is devoted to addressing the problem of overfitting, as it can arise at any noise level if model training is not carried out properly. However, the level of noise in financial data is so extreme that standard overfitting-prevention techniques, effective in many other domains, often become inadequate or insufficiently powerful in finance. To illustrate this, consider **out-of-sample testing**, a widely used practice in machine learning. In such tests, a trained model is evaluated on a new dataset that was not seen during training. In most applications, strong performance on out-of-sample data is taken as evidence that the model has learned a genuine underlying pattern. In finance, however, the data are so noisy that out-of-sample results themselves are often dominated by noise. As a consequence, even good out-of-sample performance may be purely coincidental. This means that additional validation and robustness checks are required to ensure that a model is genuinely capturing meaningful structure rather than exploiting random fluctuations.

### Limited Amount of Data

The problem of a limited amount of data is closely related to the high noise-to-signal issue discussed in the previous section. One common way to address high noise levels is to use more data: with sufficient data, noise tends to cancel out. Consider again the example of apartment prices. If we have an extremely large dataset, then for any fixed set of input values (that is, apartment characteristics), we can find many apartments with the same parameters but different prices. These prices differ because of random noise. However, if we average these prices, the random noise largely cancels out, and the resulting value should be close to the "true" price determined by the underlying relationship between the inputs and the price, that is, the price without noise.

What constitutes a sufficient amount of data depends on the level of noise. In problems with very little noise, even ten data points may be enough. In contrast, in situations with extremely high noise, even tens of millions of data points may be insufficient. As mentioned earlier, financial data, particularly prices of financial instruments, are characterized by a very high level of noise. Consequently, a large amount of data would be required to deal with the problem of noise. However, unlike in many other fields, in finance we cannot simply obtain more data than already exist. Any financial instrument, whether a stock or Bitcoin, only has price data for the period during which it has existed. We do not have access to alternative versions of our universe to observe additional realizations of how these prices evolve. In many other domains, additional data can often be generated if needed. For example, when building a machine-learning system to recognize cats and dogs in images, we can create more data by taking additional photos if we find that the dataset contains too few labeled examples.

### Dynamic Patterns

Another factor that makes machine-learning modeling in finance particularly challenging is that the patterns we seek in financial data are not only very weak but also highly dynamic. This means that such patterns may emerge, evolve or change over time, and eventually disappear. This property, like the weakness of the signal itself, has a clear underlying cause. Traders are constantly "hunting" for signals hidden in noise. When a detectable signal emerges, it is eventually discovered and exploited by market participants. As more traders begin to use the same pattern, it disappears. Because market participants typically capture the signal only approximately rather than perfectly, the signal does not vanish instantly. Instead, only the exploited portion vanishes. As a result, the signal gradually evolves before ultimately disappearing completely.

In many other fields where machine learning is applied, this issue is far less pronounced. For example, when building a system to recognize cats and dogs in images, we can safely use photographs that are several decades old, because cats and dogs still look the same. This stability makes it possible to accumulate relevant data over time, often spanning many years or even decades, without worrying that the underlying patterns have fundamentally changed.

The problem of dynamic patterns is closely intertwined with both overfitting and the limited availability of data. It is often observed that certain trading strategies perform well on training data but fail when tested on data from a different time period. One likely explanation is overfitting: the strategy may have captured noise specific to the training interval rather than a genuine, persistent signal (pattern). However, an alternative explanation is that the strategy correctly identified real patterns that existed during the training period but later disappeared or changed in the subsequent time interval in which the strategy was tested or deployed. In practice, it is often difficult to determine which explanation is correct, especially because both effects may be present simultaneously to varying degrees.

The presence of dynamic patterns makes the problem of insufficient data even more severe. As discussed earlier, we only have access to price records from the period during which a financial instrument has existed, and there is no way to obtain additional historical prices. Moreover, we cannot even be certain that all available historical data are relevant. There is no guarantee that price movements from several years ago reflect the same patterns, or even similar dynamics, as those operating in the market today. This is a serious issue, because training a model on irrelevant data can mislead it, causing the model to base its trading decisions on patterns that no longer exist. 

### Absence of Labels / Targets

In machine learning (or more specifically, in *supervised machine learning*) a model trained on a dataset is essentially a function. This function takes inputs as its arguments, combines them in a specific and potentially very complex way, and returns a value. That returned value is simply an estimate of the true outcome corresponding to the given inputs. This is exactly what we aim to do when applying machine learning to trading. More specifically, we take a set of inputs (also known as "indicators", "parameters", or "features") that are believed to have predictive power for future price movements. We then feed these inputs into our model (which is nothing but a *trading strategy*). The output is a single number indicating how much of the asset we should buy. If the number is negative, it indicates that we should sell the corresponding amount. This number is also called *allocation* or *position*.

Now comes yet another challenge. In supervised learning, we have examples of what the outputs of the function (the trained model) should be. These examples of desired outputs are included in the training dataset and are called *labels* or *targets*. During training, we (roughly speaking) try different candidate functions (models), let them produce their outputs, and measure how close those outputs are to the corresponding targets. The function that produces the best outputs, meaning those closest to the targets, is selected as the result of the training process. Thus, targets are an essential component of supervised learning. In our case, however, we do not have targets. In other words, we do not have optimal positions or allocations recorded in the dataset, and therefore we cannot train a model in the conventional supervised way. What we do have are price movements, which (intuitively) are what we would like to predict. However, even if we build a model that predicts price movements, we still need to figure out how to translate those predictions into trading decisions, that is, into concrete *positions* / *allocations* and this question is far from trivial. In this course, we present several rigorous yet practical approaches to addressing this problem.

### Unusual Measures of Quality

As briefly mentioned in the previous section, in conventional machine learning the quality of a model is measured by how close its outputs (predictions) are to the targets. For this purpose, metrics such as the mean squared error or the mean absolute error are commonly used. Sometimes more sophisticated measures are applied, but in all cases the goal is the same: to quantify how close the model’s outputs are to the desired values, known as *targets*.

In systematic trading, however, the situation is fundamentally different. Our outputs (*positions*, *allocations*) are not meant to be close to the targets (if we use *price changes* as our *targets*). Instead of computing squared or absolute differences between outputs and targets, we multiply them to calculate the profit generated by a trade. Moreover, we do not simply average these products over the dataset. Instead, we compute quantities such as the mean of these products divided by their standard deviation (a measure commonly known as the Sharpe ratio). From the perspective of conventional machine learning, these steps may appear quite unconventional.

You might think that this is not really a problem, since we are simply using a different (perhaps unusual) measure of model performance. However, this choice has at least two important disadvantages: one practical and one deep and fundamental. Let us start with the practical one. There exists a vast ecosystem of highly advanced, optimized, and powerful machine-learning libraries. Each of these libraries has been developed by large teams over many years, or even decades. They provide access to a wide range of powerful models, along with convenient tools and interfaces for model training and validation. The problem is that virtually none of these libraries natively support the unconventional performance measures used in finance, such as the Sharpe ratio or similar metrics. While it is theoretically possible to rewrite or extend these libraries to accommodate custom performance measures, doing so would be extremely time-consuming and impractical. In this course, we present a way to overcome this limitation. Using a simple but powerful trick that we will introduce, you will be able to use standard machine-learning libraries to optimize the Sharpe ratio directly! Importantly, this approach is not an approximation, it is mathematically rigorous and optimal.

Now let us turn to the second, more fundamental consequence. As mentioned earlier, in conventional (or “vanilla”) machine learning we search for a model that produces predictions as close as possible to the targets, typically by minimizing a loss function such as the mean squared error or the mean absolute error. Suppose that during training you encounter two candidate models: one that is very accurate and another that is clearly less accurate. If you were to average their outputs (predictions), the resulting model would be better than the poor model but worse than the good one. This behavior is a direct consequence of using loss functions such as mean squared or mean absolute error. Intuitively, this also makes sense: you cannot improve a good solution by mixing it with a worse one. As a result, there is no incentive to perform such averaging. Instead, you simply discard the inferior model and keep the better one. In this way, each time you find a better model, you permanently replace the previous one, continuing this process until you arrive at the best possible model.

Now, if we use the Sharpe ratio mentioned earlier, the situation changes dramatically. It turns out that when two models are combined (or, equivalently, when two trading strategies are traded together) the resulting strategy can easily outperform the better of the two individual strategies. This also makes intuitive sense. Each strategy generates its own random profit fluctuations. When the strategies are combined, these fluctuations partially cancel each other out. As a result, the combined profit stream becomes smoother and more stable, which increases the Sharpe ratio (a metric that, by definition, measures the smoothness and stability of *returns*).

If you want to take advantage of this property, you must fundamentally rethink how model training is performed. Instead of searching for the single best model among a large set of candidates (as is done in conventional machine learning) you must answer a different question: how to combine many candidate models into one powerful model (trading strategy). From a mathematical perspective, consider a multidimensional space of model parameters. In conventional machine learning, training corresponds to finding a single point in this space that represents the best-performing model. Here, however, the goal shifts from identifying a single point to identifying an optimal “cloud” (distribution, scalar field) within that space. This is a profound conceptual change and requires a substantial rethinking, and in many cases a complete redesign, of the training algorithm.
