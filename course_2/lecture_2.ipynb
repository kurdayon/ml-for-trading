# Types of Financial Data

## 1. Fundamental Data

Fundamental data refers to information about a company that comes from **official documents and internal performance measurements**. One major source is **regulatory filings**, which are formal reports that companies are legally required to submit to government authorities (such as annual and quarterly reports). Another source is **business analytics**, which includes structured analyses of a company’s operations, such as sales figures, costs, profits, and efficiency metrics used to evaluate how the business is performing.

Most fundamental data consists of **accounting data**, meaning numerical information produced by a company’s accounting system. This includes items such as revenue, expenses, assets, liabilities, and profits, which summarize the company’s financial condition and performance. These figures are usually compiled and reported on a quarterly basis.

An important characteristic of this data is that it is **reported with a time delay**, often called a *lapse*. This means the data does not become public immediately when the reporting period ends. Instead, companies first close their books, prepare the reports, and then submit and publish them days or weeks later.

Because of this delay, analysts must verify the **exact release date** of each data point and only use it in analysis after it was actually made public. A common beginner’s mistake is to assume that the data was available at the end of the reporting period (for example, on the last day of the quarter). In reality, the data is published **after the period ends**, typically when the company files its reports sometime in the following weeks.

A second important feature of fundamental data is that it can **change after it is first published**. This happens mainly in two ways: **backfilling** and **revisions**.

**Backfilling** occurs when data that was **missing** at the time is later filled in with a value. Even though the value is added later, it may appear in databases as if it had always existed, even though it was unknown at the original release date.

A **revised** (or **reinstated**) value is a correction to data that was originally published incorrectly. Companies may discover errors after releasing their results and then issue corrected numbers. These corrections can happen months or even years after the original report, and they may be revised multiple times.

Many data providers replace the original numbers with the corrected ones. This creates a problem: the corrected values were **not available** to investors or analysts at the time of the original release. Using them as if they were known at that time introduces false information into historical analysis. To address this issue, some data vendors store **multiple versions of the same data**, each with its own release date. Despite this, it is still very common to see studies that incorrectly use the **final revised value** and treat it as if it were known at the time of the first release, or even as if it were known at the end of the reporting period. This leads to unrealistic and misleading results.

Fundamental data is extremely regularized and low frequency. By *regularized*, we mean that such data is highly standardized: it is reported in fixed formats, according to well-defined accounting rules and reporting conventions, and released on predictable schedules (e.g., quarterly or annually). This standardization makes the data easy to compare across firms and time. Because of this, fundamental data is widely accessible to market participants. *Accessible* here does not necessarily mean free of charge, but rather that the data is straightforward to obtain from many sources (such as company filings, data vendors, and financial platforms) and requires little specialized infrastructure or proprietary processing to use.

Given its broad availability, fundamental data is rapidly incorporated into prices by the financial *marketplace*, meaning the collective set of investors, analysts, traders, and institutions who actively analyze information and trade financial assets. As a result, it is unlikely that fundamental data alone contains much untapped predictive value. Nevertheless, it can still be useful when combined with other data types (especially higher-frequency, less standardized, or more difficult-to-access sources) where it may provide valuable context or grounding.

Here are some examples of fundamental data:

1. **Assets**: Resources a company owns or controls (such as cash, buildings, or equipment) that are expected to provide future economic benefits.

2. **Liabilities**: Obligations a company owes to others, including loans, accounts payable, and other forms of debt.

3. **Equity (Shareholders’ Equity)**: The residual value of a company after liabilities are subtracted from assets, representing owners’ claims on the business.

4. **Revenue (Sales)**: The total income a company generates from selling goods or services during a specific period.

5. **Costs (Expenses)**: The money a company spends to operate its business, such as wages, rent, materials, and utilities.

6. **Earnings (Net Income)**: The profit remaining after all costs, expenses, interest, and taxes have been deducted from revenue.

7. **Cash Flow**: The net movement of cash into and out of a company, showing its ability to generate cash to sustain operations and growth.

8. **Profit Margins**: Ratios that show how much profit a company keeps from each unit of revenue, indicating operational efficiency.

9. **Debt Levels**: Measures of how much borrowing a company uses to finance its operations, often assessed relative to assets or equity.

10. **Macroeconomic Variables**: Economy-wide indicators such as inflation, interest rates, GDP growth, and unemployment that influence business performance and investment conditions.

## 2. Market Data

Market data refers to all information generated by **actual trading activity**. This activity takes place either on an **exchange** or on a **trading venue**. An exchange is a centralized and regulated marketplace where buyers and sellers trade standardized financial instruments. A well-known example is the **CME**, one of the world’s largest futures and derivatives exchanges. A trading venue is a broader term that also includes electronic platforms that facilitate trading without necessarily operating as traditional exchanges. One such platform is **MarketAxess**, which is widely used for trading bonds and other fixed-income instruments.

Ideally, a data provider supplies a **raw feed**, meaning the most detailed and unprocessed version of the data. A raw feed contains every event exactly as it occurred, without aggregation, filtering, or simplification. This allows analysts to reconstruct what truly happened in the market, rather than relying on summaries such as daily prices or volumes.

This raw feed often includes **unstructured information**, which refers to data that does not come in neat tables with predefined columns. Instead, it consists of detailed event messages that vary in length and content, such as order submissions, cancellations, modifications, and responses from other market participants. These messages capture the full complexity of trading behavior.

A major component of this unstructured data is **FIX messages**. FIX (**Financial Information eXchange**) is a standard electronic messaging format used by trading systems to communicate orders, trades, and related actions. FIX messages record fine-grained details such as who sent an order, what type it was, when it arrived, and how it was handled. When collected over time, FIX messages allow the reconstruction of the **trading book** (also called the **order book**), which is the continuously updated list of all buy and sell orders waiting to be executed in the market.

Another example of detailed market data is the full set of **BWICs (bids wanted in competition)**. A BWIC is a process, common in bond markets, where an investor announces an intention to sell securities and invites multiple dealers to submit competing bids. The responses reveal valuable information about market demand and pricing behavior.

Every participant in the market leaves a recognizable **footprint** in these trading records. This means that their behavior follows patterns (such as order size, timing, or execution style) that can be detected statistically. With enough data and careful analysis, it becomes possible to **anticipate a competitor’s next move**, meaning to infer what another trader or algorithm is likely to do next based on their past behavior.

All of this information (orders, executions, timestamps, and message details) is commonly referred to as **FIX data**, because it is encoded using the FIX protocol. Unlike fundamental data, FIX data is difficult to process due to its complexity, unstructured nature, and volume (it is extremely abundant, with tens of terabytes generated every day). This combination makes market data particularly challenging to use and, therefore, valuable for advanced trading strategy research.

Here are some examples of market data:

1. **Price / Yield / Implied Volatility**: The current market value of an instrument (price), its return (yield), or the market’s expectation of future volatility derived from option prices.

2. **Trading Volume**: The total number of shares, contracts, or units traded over a given period, indicating market activity and liquidity.

3. **Bid and Ask Quotes**: The best available prices at which market participants are willing to buy (bid) or sell (ask) an instrument.

4. **Spread (Bid–Ask Spread)**: The difference between the bid and ask prices, reflecting transaction costs and market liquidity.

5. **Open Interest**: The total number of outstanding derivative contracts (such as options or futures) that have not yet been closed or settled.

6. **Dividends / Coupons**: Cash payments made to shareholders (dividends) or bondholders (coupons) based on the terms of the financial instrument.

7. **Order Book (Market Depth)**: A real-time list of buy and sell orders at different price levels, showing supply and demand across the market.

8. **Quotes, Modifications, and Cancellations**: Updates to posted orders, including new quotes, changes to existing ones, or removals, which reflect trading intentions.

9. **Aggressor Side**: An indicator of whether a trade was initiated by a buyer or a seller, often inferred from whether it occurred at the ask or the bid.

10. **Time and Sales (Trade Prints)**: A chronological record of executed trades showing price, size, and time, used to analyze short-term market behavior.

## 3. Analytics Data

You can think of analytics as **derivative data**: data that does not originate directly from an underlying economic or real-world process, but is instead produced by transforming one or more original data sources. In this context, *derivative* does not mean merely “processed” in a generic sense, but rather that the data represents an **interpretation**, **aggregation**, or model-based extraction of information. For example, analytics may involve summarizing raw observations, estimating latent variables, scoring entities (*i.e.* companies) according to predefined criteria, or applying statistical or machine-learning models to infer quantities that are not directly observable. The original inputs can be (1) fundamental data, (2) market data, (3) alternative data, or even (4) other analytics.

What characterizes analytics is therefore not the type of information it conveys, but the fact that it is *constructed*: it is not directly obtainable from a primary source without applying a specific methodology. This construction embeds assumptions, parameter choices, and design decisions that shape the final output.

Analytics is sold in several different forms, depending on the provider and use case. In some cases, it takes the form of qualitative or semi-structured outputs, such as written research reports, thematic notes, or analyst opinions. In other cases, analytics is delivered as quantitative data: time series, cross-sectional scores, rankings, classifications, factor values, sentiment indicators, or probability estimates that can be directly consumed by models or trading systems. Many providers offer both, for example a numerical signal accompanied by documentation or commentary explaining its rationale.

**Investment banks** and **research firms**, for instance, sell analytics derived from in-depth analyses of companies’ business models, activities, competitive positioning, and outlook. Specialized data firms may sell analytics derived from alternative data, such as sentiment scores extracted from news articles or social media posts, foot-traffic indices inferred from location data, or sales estimates inferred from transaction records.

A key positive aspect of analytics is that the signal has already been extracted from raw inputs, saving the consumer the cost and complexity of building the processing pipeline themselves. However, there are important trade-offs. Analytics can be **expensive**, the methodology used to produce them may be partially opaque (not transparent) or proprietary, and the embedded assumptions may introduce bias. Moreover, because the same analytics are typically sold to multiple clients, any informational advantage may be diluted.

Finally, there is generally *no strict guarantee* that analytics are produced in exactly the same way over long periods of time. Providers may update models, change input data sources, refine definitions, or adjust preprocessing steps as new information or techniques become available. While reputable vendors often aim for consistency and may document major changes, smaller methodological shifts can occur without fully transparent disclosure. This lack of methodological stationarity is an important consideration when using analytics in historical analysis, backtesting, or long-term systematic strategies.

Here are some examples of analytics data:

1. **Analyst Recommendations**: Professional analysts’ opinions (such as buy, hold, or sell) based on their evaluation of a company or security.

2. **Credit Ratings**: Independent assessments of a borrower’s creditworthiness, indicating the likelihood of default on debt obligations.

3. **Earnings Expectations**: Forecasts of a company’s future earnings, typically expressed as expected earnings per share for upcoming periods.

4. **News Sentiment**: A quantitative measure of whether news coverage about an asset or company is positive, neutral, or negative.

5. **Price Targets**: Analysts’ estimates of a security’s future price level over a specified time horizon.

6. **Risk Metrics**: Statistical measures such as volatility, beta, or value-at-risk that quantify the potential downside or uncertainty of an investment.

7. **Valuation Multiples**: Ratios like price-to-earnings (P/E) or price-to-book (P/B) used to compare a company’s market value to its fundamentals.

8. **Consensus Estimates**: Aggregated forecasts from multiple analysts that represent the market’s average expectation.

9. **Factor Scores**: Composite indicators (for example value, momentum, or quality scores) derived from multiple data inputs to rank or classify securities.

10. **Event Impact Analysis**: Quantitative assessments of how specific events (such as earnings releases or policy announcements) affect prices or volatility.

## 4. Alternative Data

Alternative data can be grouped into three broad categories based on how it is generated.
- Data produced by **individuals**, such as social media posts, online news, web searches, and other forms of digital activity.
- Data generated by **business processes**, including transaction records, corporate operational data, and information collected by government agencies.
- Data produced by **sensors**, such as satellite imagery, geolocation signals, weather sensors, traffic cameras, and similar technologies. Common examples of sensor-based alternative data include satellite images used to monitor tanker movements, traffic through tunnels, or changes in parking lot occupancy.

What truly distinguishes alternative data is that it represents **primary information**, that is, information that has not yet appeared in traditional data sources. Traditional sources include financial statements, regulatory filings, analyst reports, news summaries, and market prices. Alternative data captures real-world activity **before** it is aggregated, interpreted, reported, or formally disclosed. Information may not have “made it” to those other sources simply because it takes time for raw activity to be measured, compiled, verified, and officially published. As a result, alternative data often reflects economic or business developments months earlier than fundamental data or market reactions.

For example, physical activity such as transportation flows, infrastructure usage, or production-related movements can occur long before those activities are reflected in earnings reports, analyst commentary, or asset prices. Alternative data aims to observe these early signals directly, without waiting for them to be summarized or filtered through official reporting channels.

Two major challenges associated with alternative data are **cost** and **privacy**. Collecting detailed real-world observations—especially through satellites, sensors, or large-scale monitoring systems—is expensive. In addition, the entities being observed may object to such monitoring, and there are often legitimate privacy concerns involving employees, customers, or unrelated bystanders.

Despite these challenges, alternative data offers the opportunity to work with truly unique and difficult datasets. In practice, the most valuable data is often the hardest to store, process, and analyze. A useful rule of thumb is that a dataset may be especially promising if it places significant strain on data infrastructure and engineering teams. In many cases, competitors may avoid such data because of logistical difficulties, abandon their efforts partway through, or process the data incorrectly, creating opportunities for those willing and able to handle the complexity.

Here are some examples of alternative data:

1. **Satellite / CCTV Images**: Visual data used to estimate real-world activity, such as store traffic, crop yields, or industrial production.

2. **Web Search Data (e.g., Google Searches)**: Aggregated search query trends that indicate consumer interest, demand, or emerging topics.

3. **Social Media Data (Twitter, forums, chats)**: User-generated content analyzed to gauge public opinion, sentiment, or event awareness in near real time.

4. **Mobile Location Data**: Anonymized GPS and location pings that reveal foot traffic patterns and consumer movement behavior.

5. **Credit and Debit Card Transaction Data**: Aggregated purchase records that provide insight into consumer spending trends before official sales reports.

6. **Web Traffic and App Usage Data**: Metrics on website visits or app engagement used to infer business performance or user growth.

7. **Shipping and Supply Chain Data**: Information from vessel tracking, freight flows, or customs records used to monitor trade and inventory levels.

8. **Job Postings and Hiring Data**: Online job listings analyzed to assess company expansion, labor demand, or economic momentum.

9. **Sensor and IoT Data**: Machine-generated data from physical sensors measuring activity such as energy usage, manufacturing output, or logistics flows.

10. **Metadata**: Data about data (such as timestamps, geolocation, device type, or message frequency) that adds context and enables deeper behavioral analysis.
