## What kind of trading system are we talking about?

Most modern stock, futures, and crypto exchanges use a **Central Limit Order Book** (often abbreviated **CLOB**). Think of it as a **public queue system** that matches buyers and sellers. 

## Limit Orders

The key component in the name of this trading system (*Central Limit Order Book*) is the **limit order**. In essence, the market maintains a **central book** in which all *limit orders* are recorded. The term "central" means that this book is single and shared by all participants. Everyone can see which *limit orders* are currently recorded there, everyone can submit (write) new limit orders to the book, and everyone can cancel (erase) their own *limit orders* at any time. But what exactly is a *limit order*?

There are two types of limit orders: **bid limit orders** and **ask limit orders**. You submit a *bid limit order* when you want to buy a certain amount of a traded asset and are willing to pay **no more than** a specified price. The presence of this upper price bound is the reason the order is called a *limit* order. If you want to sell a certain amount of the traded asset, you submit an *ask limit order*. In this case, you specify the **minimum price** at which you are willing to sell.

In general, a limit order (whether bid or ask) expresses an intention to buy or sell a given quantity of an asset subject to a price constraint. A limit order is therefore defined by two key parameters: the **quantity** to be traded and a **price limit** that specifies the worst acceptable execution price.

It is important to understand that limit orders are merely offers or proposals. They inform other market participants about the willingness of their owners to buy or sell a specific quantity of an asset, subject to a price constraint defined by the worst acceptable price (a single number).

Market participants naturally prefer to buy at lower prices and sell at higher prices. As a result, the price of any bid limit order (an intention to buy) is always lower than the price of any ask limit order (an intention to sell). In other words, the highest bid price (the maximum price buyers are willing to pay) is always lower than the lowest ask price (the minimum price sellers are willing to accept).
The difference between the lowest ask price and the highest bid price is called **the bid–ask spread**.

Let us examine more carefully the mechanism that ensures that the price of any **ask limit order** is always higher than the price of any **bid limit order**. Assume that this condition holds initially. Now suppose a participant attempts to place a **bid limit order** with a price higher than the lowest existing ask price. This would mean that the participant is willing to buy at a price that exceeds the minimum price at which other participants are already willing to sell.

In such a situation, placing a bid limit order makes no sense. Instead of expressing an intention to buy at some maximum price (P), the participant can immediately execute trades against the existing ask limit orders and purchase the asset at prices **lower than or equal to (P)**. For this reason, a bid limit order priced above the lowest ask would be executed immediately rather than remain in the order book.

A symmetric argument applies on the sell side. It does not make sense to place an **ask limit order** with a price lower than the highest existing bid price. If a participant is willing to sell at a minimum price (P), and there already exist buyers willing to pay more than (P), the participant can immediately sell the asset by trading against those bid orders at better prices.

As a result, limit orders that cross the opposite side of the book do not persist in the order book; they execute immediately. This mechanism ensures that bid prices remain below ask prices and that the bid–ask spread is non-negative.

## Market Orders

Bid or ask orders that are executed immediately, as introduced in the end of the previous section, have a special name: they are called **market orders**, not limit orders. The key difference is that market orders **initiate trades**. In contrast, limit orders merely express offers and therefore remain **passive**, waiting in the order book until a market order arrives and triggers a trade that satisfies the price condition specified by the limit order.

In contrast to limit orders, **market orders** specify only the **quantity** to be traded and do not include any price. This is because market orders are executed at the **best available prices** in the order book.

More precisely, when a **bid (buy) market order** for a given quantity is submitted, the system first matches it against the existing **ask limit orders** with the lowest prices. If the total quantity available at the best ask price is insufficient to fully satisfy the market order, the system continues matching against ask limit orders at the next best price, and so on, until the entire quantity is filled or available liquidity is exhausted. The same logic applies symmetrically to an **ask (sell) market order**. In this case, the system matches the order against existing **bid limit orders**, starting from the highest price and moving downward. By doing so, the system ensures that the asset is sold at the **best (highest) prices available** at the time of execution.

## Stop Market Orders

Another type of order is the **stop order**. Within this category, there are **stop market orders** and **stop limit orders**. Let us first consider stop market orders.

Consider the following scenario. You expect prices to go up, so you buy an asset. However, you are aware that there is a significant risk that prices may instead go down. If the price drops sharply while you are not monitoring the market, you could suffer a large loss. Stop market orders are designed exactly for the purpose to protect yourself against this situation.

A **sell stop market order** allows you to trigger a **market sell order** as soon as the asset trades at or below a predefined price. In simple terms, it automatically sells the asset if the price falls too far.

For example, suppose you bought an asset at a price of \\$100. If you place a sell stop market order with a stop price of $80, your market sell order will be triggered as soon as a trade occurs at \\$80 or below. Once triggered, the asset will be sold at the best available prices in the market at that moment.

This mechanism helps limit losses, although the exact execution price is not guaranteed. The exact execution price is not guaranteed because there may be an insufficient number of buy limit orders at the trigger price to fill your market sell order. As a result, buy limit orders at lower prices will be used, and you may end up selling your asset at a lower price than the trigger price.

## Stop Limit Orders

Another type of order is the **stop-limit order**, which is more flexible than the previously discussed **stop-market order**. In some respects, it resembles a stop-market order.

Let us consider **sell (ask) stop orders**. A sell stop-market order is placed at a price below the current market price. If the market price falls and reaches this **trigger (stop) price**, a market sell order is activated. As a result, the asset is sold immediately at the best prices currently available in the market. However, this “best price” is not guaranteed and may be significantly lower than the stop price. For example, the stop may be triggered at $90, but if prices continue to fall rapidly, some or all of the asset may be sold at much lower prices.

In the case of a **sell stop-limit order**, the logic is slightly more advanced. When the market price falls and reaches the trigger (stop) price, a **sell limit order** is placed at a predefined limit price, which is typically lower than the stop price. The asset will be sold only at prices equal to or higher than this limit price. If the market price falls below the limit price, the order will no longer execute.

This mechanism protects the trader from selling the asset at excessively low prices. However, it comes at the cost of losing the guarantee of execution: the order may be only partially filled or not filled at all. As a result, the trader may end up holding some or all of the asset while its price continues to fall.
