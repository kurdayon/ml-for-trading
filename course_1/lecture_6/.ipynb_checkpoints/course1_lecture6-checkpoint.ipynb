```python
import yfinance as yf
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import minimize
import pandas as pd
```

```python
df = yf.download(
    "AAPL",
    start = "1960-01-01",  
    end = "2025-08-31", 
    auto_adjust = True
    )
df = df.droplevel(level='Ticker', axis=1)
```

```python
df.rename(columns = {'Open':'o', 'High':'h', 'Low':'l', 'Close':'c', 'Volume':'v'}, inplace = True)
df = df[['v','o','h','l','c']]
df.head()
```

```python
df.loc[:, 'price_change'] = 100.0 * (df.c.shift(-1) - df.c) / df.c.shift(-1)

# range
mn_15= df.c.rolling(window = 15).min()
mx_15 = df.c.rolling(window = 15).max()
df.loc[:,'rng_15'] = (df.c - mn_15) / (mx_15 - mn_15)

# current price change (not the future one!)
cpc =  100.0 * (df.c - df.c.shift(1)) / df.c

# 15-day and 30-days means
df.loc[:,'avg_15'] = cpc.rolling(window = 15).mean()
df.loc[:,'avg_30'] = cpc.rolling(window = 30).mean()

# ratio of the two means
df.loc[:,'rat'] = df.avg_15 / df.avg_30

df.loc[:,'vol_15'] = cpc.rolling(window = 15).var()

df.loc[:,'skew_30'] = cpc.rolling(window = 30).skew()
df.loc[:,'kurt_30'] = cpc.rolling(window = 30).kurt()

feats = ['rng_15', 'avg_15', 'avg_30', 'rat', 'vol_15', 'skew_30', 'kurt_30']
df = df[feats + ['price_change']].dropna()
df.head()
```

```python
get_sharpe = lambda x : np.sqrt(252) * np.mean(x) / np.std(x)
```

```python
s = round(get_sharpe(df.price_change),2)
plt.figure(figsize = (13,5))
plt.title(f'Sharpe Ratio = {s}')
plt.ylabel('Cumulative Profit')
plt.plot(np.cumsum(df.price_change), color = 'b')
plt.show()
```

```python
feat = 'rng_15'
med = df[feat].median()
df1 = df[df[feat] <= med]
df2 = df[df[feat] >  med]
p1 = 100.0*np.mean(df1.price_change) / np.mean(np.power(df1.price_change, 2))
p2 = 100.0*np.mean(df2.price_change) / np.mean(np.power(df2.price_change, 2))
print('Positions:', p1, p2)
df.loc[:, 'position'] = np.where(df[feat] <= med, p1, p2)
df.loc[:, 'profit'] = df.price_change * df.position
```

```python
s_best = round(get_sharpe(df.profit), 3)
plt.figure(figsize = (13,4.5))
plt.title(f'Sharpe Ratio = {s_best}')
plt.ylabel('Cumulative Profit')
plt.plot(np.cumsum(df.profit), color = 'b')
plt.show()
```

```python
sharpe_ratios = []
feat = 'rng_15'
med = df[feat].median()
df1 = df[df[feat] <= med]
df2 = df[df[feat] >  med]
for _ in range(10_000):
    p1 = np.random.uniform(-5.0, 5.0)
    p2 = np.random.uniform(-5.0, 5.0)
    df.loc[:, 'position'] = np.where(df[feat] <= med, p1, p2)
    df.loc[:, 'profit'] = df.price_change * df.position
    s = get_sharpe(df.profit)
    sharpe_ratios.append(s)
```

```python
plt.figure(figsize = (12,4))
plt.hist(sharpe_ratios, bins = 100, color = 'b', alpha = 0.3)
plt.axvline(x = s_best, color = 'r')
plt.show()
```

```python
feat = 'rng_15'
def median_split(df, feat):
    out = {}
    med = df[feat].median()
    df1 = df[df[feat] <= med]
    df2 = df[df[feat] >  med]
    p1 = 100.0*np.mean(df1.price_change) / np.mean(np.power(df1.price_change, 2))
    p2 = 100.0*np.mean(df2.price_change) / np.mean(np.power(df2.price_change, 2))
    df.loc[:, 'position'] = np.where(df[feat] <= med, p1, p2)
    df.loc[:, 'profit'] = df.price_change * df.position
    s = get_sharpe(df.profit)
    return {'feat':feat, 'med':med, 'p1':p1, 'p2':p2, 'shrp':s}
```

```python
ls = []
for feat in feats:
    d = median_split(df, feat)
    ls.append(d)
sdf = pd.DataFrame(ls, columns = ['feat', 'med', 'p1', 'p2', 'shrp'])
sdf
```

```python
from scipy.stats import ks_2samp
ls = []
for feat in feats:
    med = df[feat].median()
    df1 = df[df[feat] <= med]
    df2 = df[df[feat] >  med]
    res = ks_2samp(df1.price_change, df2.price_change)   
    ls.append({'feat':feat, 'p-val':100.0*res.pvalue, 'inv_p-val':round(1.0/res.pvalue,0)})
pd.set_option('display.float_format', '{:.6f}'.format)
pd.DataFrame(ls)
```

```python
from scipy.stats import ttest_ind
ls = []
for feat in feats:
    med = df[feat].median()
    df1 = df[df[feat] <= med]
    df2 = df[df[feat] >  med]
    res = ttest_ind(df1.price_change, df2.price_change, equal_var=False)   
    ls.append({'feat':feat, 'p-val':100.0*res.pvalue, 'inv_p-val':round(1.0/res.pvalue,2)})
pd.DataFrame(ls)
```

```python
from quantile_split import QuantileSplit
```

```python
ls = []
for n in [2,4,8,16,32]:
    qs = QuantileSplit(n)
    for feat in feats:
        X = df[[feat]].values
        y = df.price_change.values
        qs.fit(X, y)
        s = np.sqrt(252.0)*qs.best_score_
        ls.append({'feat':feat, 'n':n, 'shrp':s})
res = pd.DataFrame(ls)
```
