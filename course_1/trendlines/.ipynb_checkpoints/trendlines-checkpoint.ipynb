# Trendlines

```python
import yfinance as yf

# load data
t = yf.download(
    "AAPL",
    start = "1960-01-01",  
    end = "2025-08-31", 
    auto_adjust = True
    )

# restructure data
t = t.droplevel(level='Ticker', axis=1)

# rename columns
t.rename(
    columns = {
        'Open':'o', 
        'High':'h', 
        'Low':'l', 
        'Close':'c', 
        'Volume':'v'
    }, 
    inplace = True)
```
