# What is Tabular Data?

Tabular data is a very simple concept—essentially, it’s a table of numbers. Like any table, it consists of rows and columns. Each row corresponds to an entity from a certain class of entities. For example, we might have a table where each row represents an apartment. Each column corresponds to a specific property: one column might contain the apartment’s area in square meters, another might show the year it was built, and yet another might indicate the distance to the city center. In machine learning terminology, these properties are called features.

# What is Machine Learning?

In machine learning, in addition to the table with the parameters, called features (each column is a separate feature), there is a special parameter which we call a "target". In the apartment example above, the target might be the price of the apartment. To emphasize that this parameter is special, we usually place it in the last column of the table or even take it out of the table, and represent as a stand alone column, standing near the table with the features. This parameter, called target, is special, because we want to predict it based on the values of the features.

Here, we assume that the features determine the value of the target in a stochastic manner. For example, it’s clear that the price of an apartment depends on its size, age, and distance from the city center. However, this relationship is stochastic, meaning that even with identical feature values, the price can still vary.

Let us consider a specific toy example to illustrate these concepts.

### Specific Example

Let us import the NumPy library for working with tables (matrices and tensors).

```python
import numpy as np
```

Here, we create a table with 100 rows and 4 columns, filled with random numbers sampled from a normal distribution.

```python
X = np.random.normal(size = (100, 4))
```

Let us print the first 10 rows of the table.

```python
print(X[:10,:])
```

```python
A =  1.0
B = -2.0
C =  3.0
D =  7.0
E = -4.0
```

```python
f1 = X[:,0]
f2 = X[:,1]
f3 = X[:,2]
f4 = X[:,3]

Y = (A * f1) + (B * f2) + (C * f3) + (D * f4) + E
```

```python
noise = np.random.normal(size = (100,1))

Y = Y + noise
```

```python
import random
```

```python
best_deviation = None

print('Real Parameters:', A, B, C, D, E)

for i in range(1000000):
    
    a = random.uniform(-10.0, 10.0)
    b = random.uniform(-10.0, 10.0)
    c = random.uniform(-10.0, 10.0)
    d = random.uniform(-10.0, 10.0)
    e = random.uniform(-10.0, 10.0)

    P = (a * f1) + (b * f2) + (c * f3) + (d * f4) + e

    diffs = Y - P
    deviation = np.sum(diffs * diffs)

    if best_deviation == None or deviation < best_deviation:

        best_deviation = deviation
        print(i, 'Best Found Parameters:\t', a, b, c, d, e)
```

# Loading Historical Prices

We will use the yfinance library, which provides easy access to Yahoo Finance data. It is simple to use and allows us to retrieve financial data with a single command—no registration required.

First, let’s import the library and alias it as yf:

```python
import yfinance as yf
```

Let’s load Google’s stock-price data by specifying its ticker symbol, a start date, and an end date.

```python
google = yf.download(
    "GOOGL",
    start = "2020-01-01",  
    end = "2025-08-01", 
    auto_adjust = True
    )
google = google.droplevel(level='Ticker', axis=1)
```

Let us check what columns we have in the table.

```python
list(google.columns)
```

The dataset lists four key price points for each trading day:
* Open – the price when trading begins
* Close – the price at the end of the trading session
* High – the highest price reached during the day
* Low – the lowest price reached during the day

It also includes Volume, which shows how many shares were traded that day.

Let us take a look at the first five rows of the table.
